### Data Dictionary Insights

- **Exposure (181 unique values, 0–1+):** Exposure is the fraction of a year each policy was active, so a policy with Exposure = 1.0 had twice as much time to generate claims as one with 0.5. This means raw ClaimNb is not comparable across policies; what we actually want to model is the claim rate, ClaimNb / Exposure. In a Poisson regression this is done by adding log(Exposure) to the model with its coefficient fixed at 1 — this is called an offset. It adjusts for time-at-risk without being estimated as a regular predictor. Density and Area cannot both go into a linear regression because they measure the same thing: Area is just Density cut into 6 bins (A = rural, F = urban). Including both makes the two features almost perfectly collinear, so the model cannot tell which one is driving the effect, the coefficients become unstable, and in the extreme case the regression has no unique solution. We keep only one of them — log(1 + Density) or Area, not both.
- **BonusMalus (115 unique values, 50–230):** A regulatory index, not a physical measurement. 50 = maximum safe-driver discount, 100 = baseline, above 100 = surcharge. A binary flag (BonusMalus > 100) is more useful than the raw number.
- **Density vs. Area (1,607 unique vs. 6 categories):** Both measure the same thing — Area is just Density cut into 6 bins (A = rural, F = urban). Using both creates redundancy; keep either log(Density) or Area, not both.
- **Categorical cardinality (VehGas = 2, VehBrand = 11, Region = 22):** VehGas can be encoded as 0/1. VehBrand and Region would create 33 dummy columns if one-hot encoded, so consider target encoding or grouping rare categories instead.

### Executive Defense: Why We Do NOT Log-Transform ClaimNb

The automated rule flagged ClaimNb (skew = 5.60) and recommended a log transform. We override this for two reasons. First, it is mathematically impossible: over 95% of policies have 0 claims, and log(0) = −∞, so the transform cannot be computed. Second, ClaimNb is a discrete count of accidents, not a continuous volume metric like Density, so its heavy skew is normal, not a data problem. We keep ClaimNb in raw count units and model it with Poisson regression, using log(Exposure) as an offset.

### Skewness Interpretation & Domain Override

The automated rule (|skew| > 1 → log/Box-Cox) flags several features, but the recommendation must be checked against the Business Data Dictionary:

- **ClaimNb (skew = 5.60) — DO NOT log-transform.** ClaimNb is a discrete claim count and over 95% of policies have 0 claims; log(0) is undefined and log(1+x) distorts the count structure. Keep raw units and model with Poisson / Negative Binomial regression, using Exposure as the offset.
- **Density (skew = 4.65) — log-transform is appropriate.** Density is a continuous, positive volume metric with a long right tail (rural ≈ 100 vs. Paris > 20,000 people/km²). Apply `np.log1p(Density)` before modeling.
- **BonusMalus (skew = 1.73) — do not log.** This is a regulatory index anchored at 100, not a physical quantity. A binary flag (BonusMalus > 100 = surcharge driver) is more meaningful than a transform.

**Key takeaway:** continuous volume metrics (Density) benefit from log transforms; discrete count metrics (ClaimNb) must stay in raw units.

```python
print (" ================================================== ")
print (" PEARSON MULTICOLLINEARITY AUDIT ")
print (" ================================================== ")

corr_matrix = numeric_df.corr().abs()

upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

high_corr_pairs = []
for col in upper_tri.columns:
    high_corr_features =upper_tri.index[upper_tri[col] >=0.85].tolist()
    for feat in high_corr_features:
        val = upper_tri.loc[feat, col]
        high_corr_pairs.append({
           "Feature 1":feat,
           "Feature 2":col,
           "Pearson Correlation (|r|)": round(val,4),
           "Status": "ACTION REQUIRED: Severe Multicollinearity"
        })
if high_corr_pairs:
    display(pd.DataFrame(high_corr_pairs))
else:
    print("[+] No severe collinear feature pairs (|r| >= 0.85) detected across numerical metrics.")
```

### Required Report Statements

**Chart 1 — Density**
- **Primary Statement:** The dots between 4,000 and 27,000 people/km² extend beyond the upper Tukey whisker (3,997), so they are statistical outliers. However, from a business view they are valid policyholders living in high-density cities such as Paris. Because these are real customer records rather than data entry errors, we do not drop them. Instead, we apply a log transformation (log(1 + Density)) to pull the long right tail closer to the center before modeling.
- **Alternative Conditional Statement (If Trimming):** If our business use case only targets suburban policyholders, we can trim at the upper fence (4,007 people/km²). However, if our goal is nationwide risk pricing, dropping urban policyholders would introduce selection bias.

**Chart 2 — DrivAge**
- **Primary Statement:** DrivAge shows a roughly symmetric, bell-shaped distribution centered around a median of 44 years (Q1 = 34, Q3 = 55). All values fall within valid legal driving ages (18 to 100) with no severe skewness or impossible values. No log transformation or outlier trimming is required; standard Z-score scaling in Module 2 will normalize this feature properly.
- **Alternative Conditional Statement (If Bimodal):** If the KDE curve had shown an isolated secondary bump beyond age 100, we would investigate data entry corruption. Because the KDE drops smoothly to zero at 100, the data represents a valid adult driver pool.

**Chart 3 — BonusMalus**
- **Primary Statement:** The interquartile range (Q1 = 50, Q3 = 64) shows that over 75% of policyholders hold safe-driver discounts, and the median of 50 means more than half receive the maximum discount. The upper Tukey whisker stops at 85, so drivers with scores above 85 appear as individual outlier dots extending up to 230. Because these points represent legitimate regulatory surcharges rather than data errors, we do not remove them. Instead, we recommend creating a binary flag (Is_Surcharge_Driver = BonusMalus > 100) during feature engineering.
- **Alternative Conditional Statement (If Winsorizing):** If extreme penalty scores (> 200) destabilize linear regression coefficients during initial trial runs, we can cap (winsorize) BonusMalus at the 99th percentile while retaining the binary surcharge indicator.

**Chart 4 — VehAge**
- **Primary Statement:** VehAge is heavily right-skewed, with most vehicles between 0 and 20 years old (Q1 = 2, Q3 = 11). The upper Tukey whisker stops at 24, but the maximum value reaches 100. Unlike Density and BonusMalus, a 100-year-old vehicle on an active policy is not realistic and most likely reflects a data entry placeholder rather than a real car. We recommend capping VehAge at a realistic limit (around 30 years) or removing those records, then applying log(1 + VehAge) to handle the remaining skew.
- **Alternative Conditional Statement (If Vintage Cars Are Valid):** If the business confirms that classic or vintage vehicles are a genuine customer segment, we would keep those records and create a binary flag (Is_Vintage = VehAge > 30) instead of dropping them.

```python
target_feature = "Density"

Q1 = df[target_feature].quantile(0.25)
Q3 = df[target_feature].quantile(0.75)
IQR = Q3 - Q1
iqr_lower = Q1 - 1.5 * IQR
iqr_upper = Q3 + 1.5 * IQR

df_iqr_cleaned = df[(df[target_feature] >= iqr_lower) & (df[target_feature] <= iqr_upper)]

z_scores = (df[target_feature] - df[target_feature].mean()) / df[target_feature].std()
df_z_cleaned =df[abs(z_scores) <= 3.0]

print(f"=== OUTLIER AUDIT REPORT FOR '{target_feature}' ===")
print(f"Original Record Count: {len(df):,} records")
print(f"Tukey IQR Range [{iqr_lower:.2f}, {iqr_upper:.2f}] :Excludes {len(df) - len(df_iqr_cleaned):,} records")
print(f"Z-Score Threshold (|z| <= 3.0): Excludes {len(df) - len(df_z_cleaned):,} records")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x=df[target_feature], ax=axes[0], color='salmon')
axes[0].set_title(f'BEFORE Filtering: Raw {target_feature}', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Population Density (People / km2)')

sns.boxplot(x=df_iqr_cleaned[target_feature], ax=axes[1], color='lightgreen' )
axes[1].set_title(f'AFTER Filtering: IQR Tukey Filtered {target_feature}', fontsize =12, fontweight ='bold')
axes[1].set_xlabel('Population Density (People / km2)')

plt.tight_layout()
fig_filter_path = os.path.join("..","reports", "figures", "outlier_filtering_comparison.png")
plt.savefig(fig_filter_path, dpi=300)
plt.show()
print(f"[+] Saved visual outlier comparison artifact to : {fig_filter_path}")
```

## Interpretation Statements
**1. Baseline Outlier Audit Summary:**
Our outlier audit evaluated raw population density (Density) across 678,013 policy records. The non-parametric Tukey 1.5×IQR filter set an upper fence at 4,007 people/km², identifying 77,566 records (11.44%) as statistical outliers. In contrast, the parametric Z-score filter (|Z| > 3.0) set a much higher cutoff at about 13,669 people/km², excluding only 15,209 records (2.24%).

**2. Comparative Methodological Evaluation:**
Because Density is heavily right-skewed (skewness = 4.65), the Z-score method suffers from mean distortion: the mean (1,792) is far above the median (393), and the standard deviation (3,959) is inflated by metropolitan areas like Paris, pushing the Z-score threshold unnaturally high. Tukey's IQR method relies on Q1 and Q3, which are not affected by extreme values, so it gives a robust boundary that isolates the main rural/suburban distribution. The AFTER box plot confirms this: with values above 4,007 removed, the box (92 to 1,658) and whiskers become clearly visible instead of being compressed against zero.

**3. Conditional Decision Statements:**
- **Scenario A (Nationwide Risk Pricing):** If our goal is a nationwide tariff model covering all French policyholders, we must NOT drop the 77,566 urban policyholders flagged by Tukey. Removing 11.44% of the data would introduce severe selection bias and leave the company unable to price urban risk. Action: keep all valid records and apply log(1 + Density) to remove the skew.
- **Scenario B (OLS Linear Regression):** If downstream models are limited to Ordinary Least Squares regression, which is sensitive to extreme leverage points, we recommend Tukey trimming at 4,007 people/km² to keep parameter estimates stable. This is what `src/clean_outliers.py` automates.
- **Scenario C (Tree-Based Models):** If using XGBoost, LightGBM, or Random Forest in Module 3, outlier trimming is unnecessary. Tree models split at ordinal thresholds and are naturally immune to extreme tail values.
